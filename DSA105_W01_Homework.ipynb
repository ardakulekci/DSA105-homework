- 14 textbooks at 240 TL each;
- 9 workbooks at 160 TL each.

Write the input, process, and output in your own words.

## 2. Algorithm

Write 5–7 precise steps.

## 3. Prediction

Calculate the expected revenues and total by hand before coding.


```python
textbook_unit_price = 240
textbook_quantity = 14
workbook_unit_price = 160
workbook_quantity = 9

# Complete the program using meaningful names.
textbook_revenue = textbook_unit_price * textbook_quantity
workbook_revenue = workbook_unit_price * workbook_quantity
total_revenue = textbook_revenue + workbook_revenue

print("Textbook_revenue=",textbook_revenue)
print("Workbook revenue=", workbook_revenue)
print("Total revenue=", total_revenue)
```
