## Bookstore transfer task

A bookstore sold:

- 21 fiction books at 240 TL;
- 16 non-fiction books at 180 TL.

Build a sequential program that calculates each revenue and the total.


## Algorithm and prediction

Write the algorithm first. Then predict all three outputs.


store the fiction book price and quantity
store the non-fiction book price and quantity
calculate the fiction revenue
calculate the non-fiction revenue
add the two revenues
display all three revenues

Fiction revenue = 5040
Non-fiction revenue = 2880

```python
fiction_unit_price = 240
fiction_quantity = 21
nonfiction_unit_price = 180
nonfiction_quantity = 16

# Complete the sequential program.
fiction_revenue = fiction_unit_price * fiction_quantity
nonfiction_revenue = nonfiction_unit_price * nonfiction_quantity
total_revenue = fiction_revenue + nonfiction_revenue

print("Fiction revenue =", fiction_revenue)
print("Non-fiction revenue =", nonfiction_revenue)
print("Total revenue =", total_revenue)
```
